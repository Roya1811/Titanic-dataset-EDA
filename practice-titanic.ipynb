```python
#Колонка со значением True для взрослых(старше 18 лет)
df["is_adult"] = df["Age"] > 18
df.head(10)
```

```python
df["is_adult_survived"] = df["is_adult"]
df.loc[df["Survived"] == 0, "is_adult_survived"] = False
df.head(20)
```

```python
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = (8, 5)
```

```python
#Фильтруем пропущенные значения возраста
age_data = df['Age'].dropna()
plt.hist(age_data, bins=30, color='skyblue', edgecolor='black')
plt.title('Распределение возраста пассажиров')
plt.xlabel('Возраст')
plt.ylabel('Количество пассажиров')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
# Группируем данные: количество выживших/погибших по полу
survived_by_sex = df.groupby(['Sex', 'Survived']).size().unstack()
survived_by_sex.plot(kind='bar', color=['#e74c3c', '#2ecc71'])
plt.title('Выживаемость в зависимости от пола')
plt.xlabel('Пол')
plt.ylabel('Количество человек')
plt.legend(['Погибли', 'Выжили'])
plt.xticks(rotation=0) 
plt.show()
```

```python
df = pd.read_csv('Titanic-Dataset.csv') 
print(df['Pclass'].value_counts())
```

```python
#кол-во пассажиров в каждом классе и сортируем 1, 2, 3
pclass_counts = df['Pclass'].value_counts().sort_index()

plt.pie(pclass_counts, 
        labels=['1-й класс','2-й класс','3-й класс'],
        autopct='%1.1f%%',                               
        colors=['#f1c40f', '#3498db', '#e67e22'],  
        startangle=140,                                  
        explode=(0.05, 0.05, 0.05))                      

plt.title('Распределение пассажиров по классам кают')
plt.show()
```

```python
# Точки для погибших пассажиров (Survived == 0) — красные
plt.scatter(df[df['Survived'] == 0]['Age'], df[df['Survived']==0]['Fare'], 
            alpha=0.6, color='red', label='Погибли', s=30)

# Точки для выживших пассажиров (Survived == 1) — зеленые
plt.scatter(df[df['Survived'] == 1]['Age'], df[df['Survived'] == 1]['Fare'], 
            alpha=0.6, color='green', label='Выжили', s=30)

plt.title('Зависимость стоимости билета от возраста')
plt.xlabel('Возраст')
plt.ylabel('Стоимость билета (Fare)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
